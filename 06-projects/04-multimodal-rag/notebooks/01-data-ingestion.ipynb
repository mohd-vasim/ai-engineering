```python
import os
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

class MMRagDataset:
    """Multimodal rag dataset"""

    def __init__(self):
        self.image_path = None
        self.filepath = None
        self.text_splitter = RecursiveCharacterTextSplitter()
        
        # Artifacts
        self.documents = None
        self.images = []


    def load_data(self, filepath: str, image_save_path: str = None, transform: bool = False):
        """Load data set"""

        self.image_path = image_save_path
        self.filepath = filepath

        md_text = None
        if self.image_path:
            md_text = pymupdf4llm.to_markdown(self.filepath, image_path=self.image_path, write_images=True)
            self.images = [image_path for image_path in os.listdir(self.image_path)]

        else:
            md_text = pymupdf4llm.to_markdown(self.filepath)

        self.documents = [Document(page_content=md_text, metadata={"filename": self.filepath})]
        

        if transform:
            self.documents = self.transform()

    def transform(self):
        """Transform to documents"""
        return self.text_splitter.split_documents(self.documents)

    
```

```python
dataset = MMRagDataset()
dataset.load_data(
    filepath="/home/vasim/Downloads/NIPS-2017-attention-is-all-you-need-Paper.pdf",
    image_save_path="../data/images/",
)
```

```python
class Encoder:
    def __init__(self, model_name="cohere.embed-v4:0"):
        self.model = BedrockEmbeddings(model_id=model_name)

    def encode_query(self, text: str) -> list:
        """Encode query"""

        return self.model.embed_query(text)
    
    def encode_image(self, image_path: str) -> list:
        """Encode image"""

        return self.model.embed_image(image_path)
```
