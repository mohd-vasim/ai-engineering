```python
import os
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

class MMRagDataset:
    """Multimodal rag dataset"""

    def __init__(self):
        self.image_path = None
        self.filepath = None
        self.text_splitter = RecursiveCharacterTextSplitter()
        
        # Artifacts
        self.documents = None
        self.images = []


    def load_data(self, filepath: str, image_save_path: str = None, transform: bool = False):
        """Load data set"""

        self.image_path = image_save_path
        self.filepath = filepath

        md_text = None
        if self.image_path:
            md_text = pymupdf4llm.to_markdown(self.filepath, image_path=self.image_path, write_images=True)
            self.images = [os.path.join(image_save_path, image_path) for image_path in os.listdir(self.image_path)]

        else:
            md_text = pymupdf4llm.to_markdown(self.filepath)

        self.documents = [Document(page_content=md_text, metadata={"filename": self.filepath})]
        

        if transform:
            self.documents = self.transform()

    def transform(self):
        """Transform to documents"""
        return self.text_splitter.split_documents(self.documents)

    
```

```python
dataset = MMRagDataset()
dataset.load_data(
    filepath=filepath,
    image_save_path="../data/images/",
    transform=True
)
```

```python
class Encoder:
    """Encoder to encode text and image"""
    def __init__(self, embed_model):
        self.model = embed_model

    def encode_query(self, text: str) -> list:
        """Encode query"""

        return self.model.embed_query(text)
    
    def encode_image(self, image_path: str) -> list:
        """Encode image"""

        return self.model.embed_image(image_path)
```

```python
from pymilvus import MilvusClient
import tqdm
from uuid import uuid4


class IngestDataToMilvus:
    """Ingest data into milvus database"""

    def __init__(
        self,
        embed_model,
        milvus_path: str = "../data/database/mm_rag.db",
        collection_name: str = "mm_rag_collection",
    ):
        self.dataset = MMRagDataset()
        self.encoder = Encoder(embed_model=embed_model)

        self.milvus_client = MilvusClient(uri=milvus_path)
        self.milvus_client.create_collection(
            collection_name=collection_name,
            auto_id=True,
            dimension=len(self.encoder.encode_query("hello")),
            enable_dynamic_field=True,
        )
        self.collection_name = collection_name

    def run(
        self,
        filepath: str,
        image_save_path: str = None,
    ):
        """Run the operation"""

        dataset.load_data(
            filepath=filepath, image_save_path=image_save_path, transform=True
        )

        # Embedding images
        img_embeddings = []
        for img_path in tqdm(
            dataset.images, desc="Generating embeddings for images..."
        ):
            try:
                img_embeddings.append(
                    {
                        "image_path": img_path,
                        "vector": self.encoder.encode_image(img_path),
                    }
                )
            except Exception as e:
                print(f"Error: {e}")

        print(f"Number of images embedded: {len(img_embeddings)}")

        # Embedding text
        txt_embeddings = []
        for doc in tqdm(
            dataset.documents, desc="Generating embeddings for documents..."
        ):
            try:
                txt_embeddings.append(
                    {
                        "id": uuid4(),
                        "content": doc.page_content,
                        "vector": self.encoder.encode_query(doc.page_content),
                    }
                )
            except Exception as e:
                print(f"Error: {e}")

        # Insert data into collection
        self.milvus_client.insert(
            collection_name=self.collection_name,
            data=txt_embeddings + img_embeddings,
        )
```

```python
ingest_to_milvus = IngestDataToMilvus(embed_model=embed_model)
```
