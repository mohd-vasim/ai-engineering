```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
```

```python
llm = init_chat_model("openai/gpt-oss-120b", model_provider="openrouter")
```

```python
llm.invoke("ping")
```

```python
from video_rag.postgres.search import AGENT_TOOLS

for tool in AGENT_TOOLS:
    print(f"  • {tool} — {tool}...")
print(f"\n✅ {len(AGENT_TOOLS)} tools ready")
```

```python
from video_rag.prompt import VIDEO_RAG_SYSTEM_PROMPT
from langgraph.checkpoint.memory import InMemorySaver
```

```python
checkpointer = InMemorySaver()

agent = create_agent(
    model=llm,
    system_prompt=VIDEO_RAG_SYSTEM_PROMPT,
    tools=AGENT_TOOLS,
    debug=True,
    checkpointer=checkpointer,
    name="video_rag_agent"
)
```

```python
messages = agent.invoke(input={"messages": [{"role": "user", "content": "hi"}]}, config={"configurable": {"thread_id": "123"}})
messages["messages"][-1].content
```

```python
def ask_agent(query: str):
    """Ask agent"""
    messages = agent.invoke(input={"messages": [{"role": "user", "content": query}]}, config={"configurable": {"thread_id": "123"}})
    print("\n\n")
    print(messages["messages"][-1].content)
```

```python
ask_agent("how many people have not wore helmet?")
```

```python
ask_agent("Show forklift and pedestrian near-miss events.")
```
